```python
#mean
prices = np.array([210, 225, 240, 255, 900])
print("mean is:", np.mean(prices))

even_data = np.array([1, 2, 3, 4])
print("mean(even n):", np.mean(even_data))
```

```python
#median
prices = np.array([210, 225, 240, 255, 900])
print("median is:", np.median(prices))

even_data = np.array([1, 2, 3, 4])
print("median(even n):", np.median(even_data))
```

```python
# Variance and standard diviation

scores = np.array([60, 70, 80, 90, 100])

mean = np.mean(scores)
deviation = scores - mean
sq_deviation = deviation **2
variance = np.var(scores, ddof = 1)
std_dev = np.std(scores, ddof = 1)

print("variance:", variance)
print("standard deviation:", std_dev.round(2))
```
