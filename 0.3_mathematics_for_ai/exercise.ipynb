```python
# EX 4 a
query = np.array([3, 4]) 
key   = np.array([4, 3])
total = 0
add = query + key

for i in range(len(query)):
    total += query[i] * key[i] 

print(total)    

# for dot,multi in keys:
#   val= dot[0]*multi[0] + dot[1]*multi[1]
    

keys = query @ key
print(keys)
```

```python
fig, ax=plt.subplots(figsize=(6, 4))
ax.quiver(0, 0, query[0], query[1], angles='xy', scale_units='xy', scale=1,
         color='orange', label='a = [3,4]')
ax.quiver(0, 0, key[0], key[1], angles='xy', scale_units='xy', scale=1,
         color='brown', label='a = [4,3]')
ax.quiver(0, 0, add[0], add[1], angles='xy', scale_units='xy', scale=1,
         color='green', label='a = [7,7]', linestyle='dashed')
ax.set_xlim(-1,12)
ax.set_ylim(-1,12)
```

```python
# |query||key|
srt = np.sqrt(query[0]**2+ query[1]**2)
srts = np.sqrt(key[0]**2+ key[1]**2)
# print(np.linalg.norm(srt)) 
print(srt)
print(srts)
```
