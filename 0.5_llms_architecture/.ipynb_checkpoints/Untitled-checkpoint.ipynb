###LABEL ENCODING

```python
from sklearn.preprocessing import LabelEncoder

# 2. Initialize the LabelEncoder use only with ordered data of heirarchical data
le = LabelEncoder()

# 3. Fit to the column and transform text to integers
df['Encoded_class']=le.fit_transform(df['fav-movie'])
print(df)
```

### ONE HOT ENCODING

```python

df_encoded = pd.get_dummies(df, columns=['fav-movie'],dtype="int", drop_first=True)
# Print the resulting dataframe
# print(df_encoded)
df_encoded
```

### BAG OF WORDS


```python
from sklearn.feature_extraction.text import CountVectorizer

# 1. Create a sample dataset with text reviews
datas = {"reviews": ["the food is good", "New food is terrible", "good food and good services"]}

df = pd.DataFrame(datas)

# 2. Initialize the CountVectorizer
# stop_words='english' removes useless words like 'the' and 'and' to save memory
vectorizer = CountVectorizer(stop_words='english')

# 3. Fit and transform the text column 
# This learns the unique vocabulary and counts the word occurrences 
bow_matrix = vectorizer.fit_transform(df['reviews'])

# 4. Convert the matrix into a clean Pandas DataFrame
# vectorizer.get_feature_names_out() extracts the learned vocabulary as column names 

df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out()) 
df_bow
```

### WORD EMBEDDINGS

<h1>TOKENIZERS <h/>

Whitespace tokenizer

```python
text = "Deep learning is amazing !"
tokens = text.split()
print(tokens)
```

```python
text = "Deep learning is amazing !"
tokens = text.split("e")
print(tokens)
```

###  word level tokenization that handles punctuation

```python
import re
text = "Deep learning is amazing!"
tokens = text.findall(r"\w+|[^\w\s]" text)
print(tokens)
```
