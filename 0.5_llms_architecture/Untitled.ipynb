### BAG OF WORDS


```python
from sklearn.feature_extraction.text import CountVectorizer

# 1. Create a sample dataset with text reviews
datas = {"reviews": ["the food is good", "New food is terrible", "good food and good services"]}

df = pd.DataFrame(datas)

# 2. Initialize the CountVectorizer
# stop_words='english' removes useless words like 'the' and 'and' to save memory
vectorizer = CountVectorizer(stop_words='english')

# 3. Fit and transform the text column 
# This learns the unique vocabulary and counts the word occurrences 
bow_matrix = vectorizer.fit_transform(df['reviews'])

# 4. Convert the matrix into a clean Pandas DataFrame
# vectorizer.get_feature_names_out() extracts the learned vocabulary as column names 

df_bow = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out()) 
df_bow
```

###this is a step by step implementation of a simple tokenizer

```python
with open ("text.txt", "r", encoding="utf-8") as f:
    raw_text = f.read()
len(raw_text),raw_text
```

### text cleaning using regular expressions

```python
white_space_splitting = re.split(r'\s', raw_text)
print(white_space_splitting)
```

```python
len(white_space_splitting)
```

```python
raw_text = re.sub(r'@\w+',"", raw_text, flags=re.MULTILINE)
raw_text = re.sub(r'#\w+',"", raw_text, flags=re.MULTILINE)
raw_text = re.sub(r'[^a-zA-Z\s]',"", raw_text)
raw_text = re.sub(r'\s+',' ',raw_text).strip()

print(raw_text)
```

```python
#splitting using other criterias
white_space_splitting_with_special_xters = re.split(r'--r|[@^!,.#*?&]|\s', raw_text)
print(white_space_splitting_with_special_xters)
```

```python
## removing whitespaces
stripe_text = [word.strip() for word in white_space_splitting_with_special_xters if word.strip()]
len(stripe_text)
```

```python
print(stripe_text)
```

```python
sorted_strip_text = sorted(set(stripe_text))
print(sorted_strip_text)
```

```python
len(sorted_strip_text)
```

```python
### ENUMERTE
for idx,token in enumerate(sorted_strip_text):
    print(f"({idx}, {token})")
```

```python
vocab = {}
for idx,token in enumerate(sorted_strip_text):
    vocab[token] = idx
(vocab)  
```

```python
id_to_str = {}
for idx,token in enumerate(vocab):
    id_to_str[idx] = token
id_to_str
```

```python
token_ids = [vocab[word] for word in stripe_text]
print(token_ids)
```

```python
tokens_recoverd = [id_to_str[idx] for idx in token_ids]
" ".join(tokens_recoverd)
```
