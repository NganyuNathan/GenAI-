```python
#CONDITIONALS
age = input("enter your age")
age_to_number=int(age)
if age_to_number > 20:
    print("age is above 20")
print("completed")   
```

```python
#if else condition
height = input("enter your height")
new_height=int(height)
if new_height >=23:
    print("tall")
    print("indeed your height is above 23")
else:
    print("you are short")

print("thanks for using our program")

```

```python
score = input("enter your score ?")
new_score=int(score)

if new_score >=90:
    print("you have an A+ grade")
    
elif new_score >=80 and new_score <=89:
    print("you have an A grade")
    
elif new_score >=70 and new_score<=79:
    print("you have a B+ grade")
elif new_score >=60 and new_score<=69:
    print("you have a B grade")
elif new_score >=50 and new_score<=59:
    print("you have a C grade") 
    
else:
    print(" sorry you have failed")
```
