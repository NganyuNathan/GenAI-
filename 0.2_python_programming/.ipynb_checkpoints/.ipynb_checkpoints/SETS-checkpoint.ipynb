<h1>SETS<h1/>

```python
numbers = {1,2,3,4}
colors = {"red", "green","blue"}
empty = set()
print(empty)
```

```python
#Methods in sets

fruits = {"apple","bananana"}
fruits.add("mango")
fruits.update(["cherry","berries",30])
fruits.remove("cherry")
fruits.discard(30)
print(fruits)
```

```python
#pop removes an returns a random item because sets are unordered
items = {"x","y"}
print(items.pop())
```

Union and Intersection

```python
a = {1,2,3,4,5}
b = {7,8,9,5,4}
print(a|b)   #union
print(a&b)  #intersection
```

```python
#difference method it is what is found in a but not in b
a={1,2,3,4}
b={5,6,7,8,3}

print(a-b)
```

```python
#symmetric is items that are in one set but not both
print(a^b)
```

SET COMPREHENSION

```python
#sqaure of numbers
sqn = []
for i in range(1,11):
    print(f"{i*i}")
    sqn.append(i*i)
print(sqn)    
    
```

```python
sqr = [i*i for i in range(1,11)]
print(sqr)
```
