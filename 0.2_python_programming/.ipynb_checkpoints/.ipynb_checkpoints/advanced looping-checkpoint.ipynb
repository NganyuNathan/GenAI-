```python
teams = ["barca", "arsenal","man city"]
pos = [1,2,1]
for item in zip(teams,pos):
    print(item)
```

```python
for team,position in zip(teams,pos):
    print(f" {team} will finish at position: {position} this season\n")
```

ENUMERATE

```python
for items in enumerate(teams):
    print(items)
```
