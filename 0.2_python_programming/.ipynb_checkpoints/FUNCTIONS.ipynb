<h2>*args vs Kwargs</h2>

```python
# args: this parameter takes in any number of arguments
def total(*values):
    t = sum(values)
    return t
total(1,2,3,4,5)   
```

```python
#**kwargs: this parameter takes in any number of keywords arguments(e.g = 10)
           # and stores them in a dictionary

def machines (**specs):
    return specs
machines(color="black", ram="16gb")    
```

```python
def machines (**specs):
    for i in specs.items():
        print(i)
    return specs
machines(x=1,y=2,z=3) 
```

<h2>Lambda functions</h2>

```python
#a function to square a number

square = lambda x : x**2
square(4)
```

```python
premier_league = [("man-u", 10), ("arsenal", 7) ,("man c", 24)]
premier_league.sort(key=lambda m: m[1], reverse=True)
premier_league
```

```python
greet = lambda name : f'hello {name}'
greet("peter")
```
