<h1> Dictionaries <h1>

```python
#creating a dictionary
profile = {
    "name":"Tom",
    "age":23,
    "hobbies":["sleeping","playing football","reading books"],
    "height":1.80
}
profile["hobbies"]
# profile.get("name")
```

```python
profile.get("country")
profile
```

```python
#Updating a dictionary
#1 adding/updating
profile["country"] = "afghanistan"
profile
#2 deleting
del profile["age"]
profile
profile.pop["height"]
profile
```

```python
profile
```

```python
#getting keys and values and values
profile.keys()
profile.values()
profile.items()
```

```python
setting = {"theme": "light", "font":"arabic"}
new_font = "san-siro"


print(setting.get("font","arial"))
setting.setdefault("cameroon","bamenda")
setting.update(new_font)
setting
```

```python
# looping over a dictionary
# -looping by keys
for key in setting.keys():
    print(key)
for value in setting.values():
    print(value)
for items in setting.items():
    print(items)

```

```python
counts = "God is the answer"
sentence = counts.split()
list_of_words = sentence.split()
for word in sentence:
    print(word)
print(list_of_words) 

        
```

```python
word_count = {}
for word in counts:
    if word in word_count.keys():
        word_count[word] = word_count[word] + 1
    else:
        word_count[word] = 1
print(word_count)        
```
