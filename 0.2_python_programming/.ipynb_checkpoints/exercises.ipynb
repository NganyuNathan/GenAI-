```python
for multi in range(0,101):
    if multi%5==0 and multi%3 == 0:
        print("fizzbuzz")
    elif multi%3 == 0:
        print("fizz")
    elif multi%5 == 0:
        print("buzz")
    else:
        print(f"{multi} is not a multiple of either 3 or 5")
```
